🔍 Objective

To evaluate how diverse the workforce is across gender, race, region, and marital status, and track the representation of diverse groups in leadership and technical roles. This analysis helps ensure fairness, equity, and inclusiveness in hiring and promotion practices.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
file_path = r"../../data/processed/cleaned3.csv"
df = pd.read_csv(file_path)
```

Univariate Distribution for Demographic Variables

```python
# Gender Distribution
sns.countplot(data=df, x='GenderCode', palette='Set2')
plt.title("Gender Distribution")
plt.show()

# Race Distribution
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='RaceDesc', palette='Set3')
plt.title("Race/Ethnicity Distribution")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Region Distribution
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='Region', palette='coolwarm')
plt.title("Employee Distribution by Region")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Marital Status
sns.countplot(data=df, x='MaritalDesc', palette='pastel')
plt.title("Marital Status Distribution")
plt.xticks(rotation=45)
plt.show()
```

Diversity in Leadership and Technical Roles

```python
# Leadership: filter by high seniority level
leadership_roles = df[df['SeniorityLevel'].str.contains("Manager|Director|VP", na=False)]

# Technical roles: filter by job family
technical_roles = df[df['JobFamily'].str.contains("Engineering|IT|Data|Technology", na=False)]

# Gender in leadership
sns.countplot(data=leadership_roles, x='GenderCode', palette='Set1')
plt.title("Gender Representation in Leadership Roles")
plt.show()

# Race in leadership
plt.figure(figsize=(10, 5))
sns.countplot(data=leadership_roles, x='RaceDesc', palette='Set2')
plt.title("Race Representation in Leadership Roles")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Gender in technical roles
sns.countplot(data=technical_roles, x='GenderCode', palette='Accent')
plt.title("Gender Representation in Technical Roles")
plt.show()

# Race in technical roles
plt.figure(figsize=(10, 5))
sns.countplot(data=technical_roles, x='RaceDesc', palette='Paired')
plt.title("Race Representation in Technical Roles")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

📈 What We Analyzed and Why

| Analysis Area                                     | Description                                   | Why It's Important                                         |
| ------------------------------------------------- | --------------------------------------------- | ---------------------------------------------------------- |
| Gender, Race, Region, Marital Status Distribution | General diversity snapshot of workforce       | Ensure balanced representation and identify potential bias |
| Gender & Race in Leadership                       | Measure diversity in decision-making roles    | Key to equity and inclusive leadership                     |
| Gender & Race in Technical Roles                  | Assess diversity in skilled, high-demand jobs | Diversity in STEM roles is often lacking and needs focus   |


| Category Analyzed    | Why It Was Analyzed                                                           |
| -------------------- | ----------------------------------------------------------------------------- |
| **GenderCode**       | To check gender balance and identify gender gaps                              |
| **RaceDesc**         | To evaluate ethnic diversity and highlight over/underrepresentation           |
| **Region**           | To understand geographic diversity and regional hiring equity                 |
| **MaritalDesc**      | To check for diversity in personal backgrounds and benefits planning          |
| **Leadership Roles** | To measure diversity among decision-makers (Managers, Directors, VPs)         |
| **Technical Roles**  | To assess inclusion in skilled, high-demand job roles (Engineering, IT, Data) |


📊 Why These Graphs Were Used

| Graph                                      | Purpose                                             | Reason for Choice                                       |
| ------------------------------------------ | --------------------------------------------------- | ------------------------------------------------------- |
| `countplot` (Gender/Race/etc.)             | Simple count of employees in each demographic group | Clear visual breakdown of representation                |
| `countplot` (Leadership & Technical Roles) | Drill-down into specific high-impact roles          | Highlights underrepresentation or bias within subgroups |


📈 Visualizations Used and Their Purpose:

| Graph                                         | Purpose                                         | Reason for Choosing It                         |
| --------------------------------------------- | ----------------------------------------------- | ---------------------------------------------- |
| `countplot` – GenderCode                      | Show overall gender distribution                | Simple, clear way to compare group sizes       |
| `countplot` – RaceDesc                        | Show overall racial/ethnic representation       | Highlights workforce balance across categories |
| `countplot` – Region                          | Analyze geographic representation               | Helps identify diversity across locations      |
| `countplot` – MaritalDesc                     | Understand diversity in social backgrounds      | Useful for benefits planning and inclusion     |
| `countplot` – Gender/Race in Leadership Roles | Focus on decision-making diversity              | Critical to assess equity in promotions        |
| `countplot` – Gender/Race in Technical Roles  | Focus on diversity in high-demand skilled roles | Helps address gaps in STEM fields              |


🧠 Why This Analysis Matters:

- Promotes fairness and equity in the workplace.

- Helps detect underrepresented groups, especially in influential roles.

- Guides HR in building inclusive hiring and promotion strategies.

- Supports diversity goals and compliance with legal frameworks (e.g., equal opportunity).

- Creates a stronger employer brand for attracting diverse talent.

💡 Business Insights That Can Be Drawn:

- Is the leadership team reflective of the overall workforce?

- Are women and minority groups fairly represented in technical or managerial positions?

- Do certain regions or departments show a lack of diversity?

- Is there evidence of unconscious bias in senior role appointments?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# --- Load Data ---
file_path = r"../../data/processed/cleaned3.csv"
df = pd.read_csv(file_path)

# --- Seaborn Styling ---
sns.set(style="whitegrid", font_scale=1.0)
plt.rcParams['axes.labelweight'] = 'bold'

# --- Define Subsets ---
leadership_roles = df[df['SeniorityLevel'].str.contains("Manager|Director|VP", case=False, na=False)]
technical_roles = df[df['JobFamily'].str.contains("Engineering|IT|Data|Technology", case=False, na=False)]

# --- Create Figure Layout ---
fig, axes = plt.subplots(4, 2, figsize=(16, 18))
fig.suptitle("Diversity & Inclusion Dashboard — Workforce Representation Overview", fontsize=16, fontweight='bold', y=0.93)

# --- (1) Gender Distribution ---
sns.countplot(data=df, x='GenderCode', palette='Set2', ax=axes[0, 0])
axes[0, 0].set_title("Overall Gender Distribution", fontsize=12, fontweight='bold')
axes[0, 0].set_xlabel("Gender")
axes[0, 0].set_ylabel("Count")

# --- (2) Race Distribution ---
sns.countplot(data=df, x='RaceDesc', palette='Paired', ax=axes[0, 1])
axes[0, 1].set_title("Overall Race/Ethnicity Distribution", fontsize=12, fontweight='bold')
axes[0, 1].set_xlabel("Race/Ethnicity")
axes[0, 1].set_ylabel("Count")
axes[0, 1].tick_params(axis='x', rotation=45)

# --- (3) Leadership Diversity by Gender ---
leadership_gender = leadership_roles['GenderCode'].value_counts()
axes[1, 0].pie(
    leadership_gender,
    labels=leadership_gender.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=sns.color_palette("Set3", len(leadership_gender))
)
axes[1, 0].set_title("Leadership Roles — Gender Composition", fontsize=12, fontweight='bold')

# --- (4) Leadership Diversity by Race ---
leadership_race = leadership_roles['RaceDesc'].value_counts()
axes[1, 1].pie(
    leadership_race,
    labels=leadership_race.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=sns.color_palette("coolwarm", len(leadership_race))
)
axes[1, 1].set_title("Leadership Roles — Race Composition", fontsize=12, fontweight='bold')

# --- (5) Technical Roles by Gender ---
sns.countplot(data=technical_roles, x='GenderCode', palette='Accent', ax=axes[2, 0])
axes[2, 0].set_title("Technical Roles — Gender Distribution", fontsize=12, fontweight='bold')
axes[2, 0].set_xlabel("Gender")
axes[2, 0].set_ylabel("Count")

# --- (6) Technical Roles by Race ---
sns.countplot(data=technical_roles, x='RaceDesc', palette='Set1', ax=axes[2, 1])
axes[2, 1].set_title("Technical Roles — Race Distribution", fontsize=12, fontweight='bold')
axes[2, 1].set_xlabel("Race/Ethnicity")
axes[2, 1].set_ylabel("Count")
axes[2, 1].tick_params(axis='x', rotation=45)

# --- (7) Regional Diversity (Gender %) ---
region_gender = pd.crosstab(df['Region'], df['GenderCode'], normalize='index') * 100
region_gender.plot(kind='barh', stacked=True, colormap='Set2', ax=axes[3, 0], edgecolor='black')
axes[3, 0].set_title("Regional Workforce Composition by Gender (%)", fontsize=12, fontweight='bold')
axes[3, 0].set_xlabel("Percentage")
axes[3, 0].set_ylabel("Region")
axes[3, 0].legend(title='Gender')

# --- (8) Regional Diversity (Race %) ---
region_race = pd.crosstab(df['Region'], df['RaceDesc'], normalize='index') * 100
region_race.plot(kind='barh', stacked=True, colormap='tab20c', ax=axes[3, 1], edgecolor='black')
axes[3, 1].set_title("Regional Workforce Composition by Race (%)", fontsize=12, fontweight='bold')
axes[3, 1].set_xlabel("Percentage")
axes[3, 1].set_ylabel("Region")
axes[3, 1].legend(title='Race', bbox_to_anchor=(1.05, 1), loc='upper left')

# --- Final Layout Adjustments ---
plt.tight_layout(rect=[0, 0, 1, 0.95])

# --- Save High-Resolution Image for Report ---
plt.savefig("../../reports/figures/diversity_inclusion_dashboard.png", dpi=300, bbox_inches='tight')
plt.show()
```
