✅ Objective:

To estimate Training ROI using performance trends over time (TenureDays) as a proxy. We'll assume that employees with longer tenure have had more exposure to training, mentoring, or learning opportunities, and should ideally show higher performance.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
file_path = r"../../data/processed/cleaned3.csv"
df = pd.read_csv(file_path)

# Convert TenureDays into TenureYears
df['TenureYears'] = df['TenureDays'] / 365
```

Scatter Plot – Performance vs. Tenure (Progression Insight)

```python
plt.figure(figsize=(10, 5))
sns.scatterplot(data=df, x='TenureYears', y='Current Employee Rating', hue='IsActive', palette='coolwarm')
plt.title("Performance Progression Over Tenure")
plt.xlabel("Tenure (Years)")
plt.ylabel("Performance Rating")
plt.grid(True)
plt.tight_layout()
plt.show()
```

Binned Tenure Groups (Optional Grouping)

```python
# Create tenure bins
df['TenureGroup'] = pd.cut(df['TenureYears'], bins=[0, 1, 3, 5, 10, 20], labels=['<1 yr', '1–3 yrs', '3–5 yrs', '5–10 yrs', '10+ yrs'])

# Average performance by tenure group
group_perf = df.groupby('TenureGroup')['Current Employee Rating'].mean().reset_index()

# Plot bar chart
plt.figure(figsize=(8, 5))
sns.barplot(data=group_perf, x='TenureGroup', y='Current Employee Rating', palette='viridis')
plt.title("Average Performance Rating by Tenure Group")
plt.xlabel("Tenure Group")
plt.ylabel("Average Performance Rating")
plt.tight_layout()
plt.show()
```

Optional – Track ROI by Job Family or Seniority

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='JobFamily', y='Current Employee Rating', hue='TenureGroup', palette='cool')
plt.title("Performance vs Tenure across Job Families")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

📈 What We’re Doing and Why

| Step                                    | What We’re Analyzing                                    | Why It Matters                                                     |
| --------------------------------------- | ------------------------------------------------------- | ------------------------------------------------------------------ |
| `scatterplot` of Performance vs Tenure  | Is there an upward trend over time?                     | Indicates whether performance improves with experience/training    |
| `barplot` of Tenure Group vs Avg Rating | Do longer-tenured employees have higher average scores? | A simple ROI proxy: higher tenure = more learning = better results |
| `boxplot` by JobFamily & TenureGroup    | Are training effects consistent across departments?     | Helps identify where training (formal or informal) is working best |


| Metric / View                            | What It Shows                                             | Why It Matters                                                    |
| ---------------------------------------- | --------------------------------------------------------- | ----------------------------------------------------------------- |
| **Performance vs. Tenure (scatterplot)** | Trends of individual performance with time spent          | If performance rises with tenure, it reflects learning and growth |
| **Average Performance by Tenure Group**  | Group-level analysis of performance by experience bracket | Helps generalize training impact across the organization          |
| **Performance vs. Tenure by JobFamily**  | Role-specific trends                                      | Identifies where training is most/least effective                 |


📈 Graphs Used and Their Purpose

| Graph Type                                       | Description                                               | Reason for Use                                          |
| ------------------------------------------------ | --------------------------------------------------------- | ------------------------------------------------------- |
| `scatterplot` – Tenure vs Performance            | Shows the individual distribution of ratings over time    | Detects trendlines or performance clustering            |
| `barplot` – Avg Rating by TenureGroup            | Binned comparison of performance across experience levels | Simplifies trend interpretation for decision-making     |
| `boxplot` – JobFamily vs Rating (by TenureGroup) | Examines if training impact varies by department          | Helps pinpoint where learning translates to performance |


🧠 Insights and Interpretations

- A visible upward trend between tenure and performance indicates that employees are improving over time, suggesting effective informal learning or training.

- Employees with 3+ years of tenure tend to show higher average ratings, implying a return on internal development efforts.

- Job families such as IT, Engineering, or Operations may show more noticeable performance gains, indicating strong skill development over time.

- Some departments with flat trends may require targeted formal training or reskilling programs.

- Provides a baseline for evaluating learning effectiveness over time.

- Supports HR in identifying high-impact areas for additional training or development.

- Enables data-driven planning for mentorship, onboarding, and learning programs.
