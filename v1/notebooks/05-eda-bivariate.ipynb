```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
file_path = r"../data/processed/cleaned3.csv"
df = pd.read_csv(file_path)

# Seaborn style
sns.set(style='whitegrid')
plt.rcParams["figure.figsize"] = (10, 5)

# 1. Barplot: Average TenureDays per Performance Score
plt.figure()
sns.barplot(data=df, x='Performance Score', y='TenureDays', estimator='mean', palette='Set2')
plt.title('Average Tenure (Days) per Performance Score')
plt.ylabel('Average Tenure (Days)')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

# 2. Violin plot: Age distribution by SeniorityLevel
plt.figure()
sns.violinplot(data=df, x='SeniorityLevel', y='Age', palette='coolwarm')
plt.title('Age Distribution by Seniority Level')
plt.tight_layout()
plt.show()

# 3. KDE Curve: Age distribution by Performance Score
plt.figure()
for score in df['Performance Score'].unique():
    subset = df[df['Performance Score'] == score]
    sns.kdeplot(subset['Age'], label=score, fill=True)
plt.title('Age Distribution by Performance Score')
plt.xlabel('Age')
plt.legend()
plt.tight_layout()
plt.show()

# 4. Barplot: Average Employee Rating per Seniority Level
plt.figure()
sns.barplot(data=df, x='SeniorityLevel', y='Current Employee Rating', estimator='mean', palette='Set1')
plt.title('Average Employee Rating by Seniority Level')
plt.ylabel('Avg Rating')
plt.tight_layout()
plt.show()

# 5. Countplot: JobFamily distribution
plt.figure()
sns.countplot(data=df, x='JobFamily', palette='pastel', order=df['JobFamily'].value_counts().index)
plt.title('Employee Count by Job Family')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 6. Countplot: Gender Distribution by Business Unit
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='BusinessUnit', hue='GenderCode', palette='Set3')
plt.title('Gender Distribution across Business Units')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 7. Barplot: Average Age by Marital Status
plt.figure()
sns.barplot(data=df, x='MaritalDesc', y='Age', estimator='mean', palette='Blues')
plt.title('Average Age by Marital Status')
plt.tight_layout()
plt.show()
```
