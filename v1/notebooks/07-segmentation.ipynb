```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
file_path = r"../data/processed/cleaned3.csv"
df = pd.read_csv(file_path)

# Style
sns.set(style='whitegrid')
plt.rcParams["figure.figsize"] = (10, 6)

# -----------------------------
# 1. Active vs Terminated Employees
# -----------------------------
plt.figure()
status_counts = df['IsActive'].value_counts()
labels = ['Active', 'Terminated']
colors = ['#4CAF50', '#FF6F61']
plt.pie(status_counts, labels=labels, autopct='%1.1f%%', startangle=140, colors=colors, wedgeprops=dict(width=0.5))
plt.title('🧑‍💼 Active vs Terminated Employees')
plt.tight_layout()
plt.show()

# -----------------------------
# 2. High Performers (Rating ≥ 4)
# -----------------------------
high_perf = df[df['Current Employee Rating'] >= 4]

plt.figure()
sns.countplot(data=high_perf, x='SeniorityLevel', palette='YlGn')
plt.title('🌟 High Performers (Rating ≥ 4) by Seniority Level')
plt.xlabel('Seniority Level')
plt.ylabel('Count of High Performers')
plt.tight_layout()
plt.show()

# -----------------------------
# 3. Entry vs Executive Level Trends
# -----------------------------
# Filter only Entry and Executive level
entry_exec = df[df['SeniorityLevel'].isin(['Entry', 'Executive'])]

# Compare Average Age
plt.figure()
sns.barplot(data=entry_exec, x='SeniorityLevel', y='Age', estimator='mean', palette='Set2')
plt.title('🧓 Avg Age: Entry vs Executive')
plt.ylabel('Average Age')
plt.tight_layout()
plt.show()

# Compare Average Tenure
plt.figure()
sns.barplot(data=entry_exec, x='SeniorityLevel', y='TenureDays', estimator='mean', palette='Set1')
plt.title('📆 Avg Tenure Days: Entry vs Executive')
plt.ylabel('Average Tenure (Days)')
plt.tight_layout()
plt.show()

# Gender Distribution in Entry vs Executive
plt.figure()
sns.countplot(data=entry_exec, x='SeniorityLevel', hue='GenderCode', palette='pastel')
plt.title('👥 Gender Split: Entry vs Executive')
plt.tight_layout()
plt.show()
```
