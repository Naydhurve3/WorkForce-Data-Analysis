```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Optional: Set plot style
sns.set(style="whitegrid")
```

```python
file_path = r"../data/processed/cleaned3.csv"
df = pd.read_csv(file_path)
```

```python
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='JobFamily', order=df['JobFamily'].value_counts().index, palette='viridis')
plt.title('Distribution of Job Family')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='SeniorityLevel', order=df['SeniorityLevel'].value_counts().index, palette='coolwarm')
plt.title('Distribution of Seniority Level')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x='RaceDesc', order=df['RaceDesc'].value_counts().index, palette='Set2')
plt.title('Race Distribution')
plt.tight_layout()
plt.show()
```

Distribution Plots (Numerical Data)

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['Age'], kde=True, bins=30, color='teal')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['TenureDays'], kde=True, bins=30, color='purple')
plt.title('Tenure (Days) Distribution')
plt.xlabel('Tenure Days')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Current Employee Rating', palette='Blues')
plt.title('Current Employee Rating Distribution')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# 🔹 Load the cleaned dataset
file_path = r"../data/processed/cleaned3.csv"
df = pd.read_csv(file_path)

# -----------------------------------------------
# 🔸 Countplots for Categorical Columns
# -----------------------------------------------
categorical_cols = ['JobFamily', 'SeniorityLevel', 'RaceDesc']

for col in categorical_cols:
    plt.figure(figsize=(10, 5))
    ax = sns.countplot(data=df, x=col, palette='Set2', order=df[col].value_counts().index)
    plt.title(f"Count of {col}")
    plt.xticks(rotation=45)
    for p in ax.patches:
        height = p.get_height()
        ax.annotate(f'{height}', (p.get_x() + p.get_width() / 2., height), 
                    ha='center', va='bottom', fontsize=10)
    plt.tight_layout()
    plt.show()

# -----------------------------------------------
# 🔸 Distribution plots for Numerical Columns
# -----------------------------------------------
numerical_cols = ['Age', 'TenureDays', 'Current Employee Rating']

for col in numerical_cols:
    plt.figure(figsize=(10, 5))
    sns.histplot(data=df, x=col, kde=True, color='skyblue', bins=30)
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()
```
