# 02 - Attrition Analysis
Explore termination patterns: rates by department, demographics, tenure, and performance.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_parquet('data/processed/workforce_clean_base.parquet')
df['is_terminated'] = df['EmployeeStatus'].str.lower().isin(['terminated', 'voluntary termination', 'involuntary termination']).astype(int)
print(f'Overall attrition rate: {df["is_terminated"].mean():.2%}')
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
df.groupby('DepartmentType')['is_terminated'].mean().sort_values().plot(kind='barh', ax=axes[0,0], title='Attrition Rate by Department')
df.groupby('GenderCode')['is_terminated'].mean().sort_values().plot(kind='barh', ax=axes[0,1], title='Attrition Rate by Gender')
df.groupby('Generation')['is_terminated'].mean().sort_values().plot(kind='barh', ax=axes[1,0], title='Attrition Rate by Generation')
df.groupby('Performance Score')['is_terminated'].mean().sort_values().plot(kind='barh', ax=axes[1,1], title='Attrition Rate by Performance')
plt.tight_layout()
```

```python
fig, ax = plt.subplots(figsize=(10, 4))
df.groupby('TenureBucket')['is_terminated'].mean().plot(kind='bar', ax=ax, title='Attrition Rate by Tenure Bucket')
ax.set_ylabel('Attrition Rate')
plt.tight_layout()
```

```python
print('\n=== Attrition by Termination Type (top categories) ===')
term = df[df['is_terminated'] == 1]
if 'TerminationType' in term.columns:
    print(term['TerminationType'].value_counts())
if 'TerminationDescription' in term.columns:
    print(f'\nSample termination descriptions:')
    for desc in term['TerminationDescription'].dropna().sample(min(5, len(term))):
        print(f'  - {desc[:100]}')
```
