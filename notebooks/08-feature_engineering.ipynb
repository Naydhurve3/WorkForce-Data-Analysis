# 08 - Feature Engineering
Review feature transformations applied: demographic, categorical, temporal, and embedding features.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from src.wf_analysis.features.demographic import DemographicFeatureTransformer
from src.wf_analysis.features.categorical import CategoricalFeatureTransformer
from src.wf_analysis.features.temporal import TemporalFeatureTransformer

df_raw = pd.read_csv('data/raw/employee_data.csv')
df_processed = pd.read_parquet('data/processed/workforce_clean_base.parquet')
print(f'Raw columns: {len(df_raw.columns)} -> Processed columns: {len(df_processed.columns)}')
print(f'\nOriginal columns:\n{list(df_raw.columns)}')
print(f'\nNew columns added:\n{list(set(df_processed.columns) - set(df_raw.columns))}')
```

```python
# Demographic features
demo = DemographicFeatureTransformer()
df_demo = demo.fit_transform(df_raw)
demo_cols = [c for c in df_demo.columns if c not in df_raw.columns]
print(f'Demographic features added: {demo_cols}')
print(df_demo[demo_cols].head())
```

```python
# Categorical features (YAML-driven mappings)
cat = CategoricalFeatureTransformer(mapping_path='config/categorical_mappings.yaml')
df_cat = cat.fit_transform(df_raw)
cat_cols = [c for c in df_cat.columns if c not in df_raw.columns]
if cat_cols:
    print(f'Categorical features added: {cat_cols}')
    print(df_cat[cat_cols].head())
```

```python
# Temporal features
temp = TemporalFeatureTransformer()
df_temp = temp.fit_transform(df_raw)
temp_cols = [c for c in df_temp.columns if c not in df_raw.columns]
print(f'Temporal features added: {temp_cols}')
print(df_temp[temp_cols].head())
```

```python
print('\n=== Processed Dataset Summary ===')
print(f'Shape: {df_processed.shape}')
print(f'\nMissing values after pipeline:')
missing = df_processed.isnull().sum()
print(missing[missing > 0].to_string() if missing.any() else 'None')
```
