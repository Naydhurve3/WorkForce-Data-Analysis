# 07 - Workforce Planning
Tenure distributions, headcount trends, turnover risk, and succession planning indicators.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_parquet('data/processed/workforce_clean_base.parquet')
df['is_terminated'] = df['EmployeeStatus'].str.lower().isin(['terminated', 'voluntary termination', 'involuntary termination']).astype(int)
print(f'Active employees: {(df["is_terminated"] == 0).sum()}')
print(f'Terminated employees: {df["is_terminated"].sum()}')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
df['TenureYears'].plot(kind='hist', bins=20, ax=axes[0], title='Tenure Distribution')
axes[0].set_xlabel('Years')
df['TenureBucket'].value_counts().plot(kind='bar', ax=axes[1], title='Tenure Buckets')
plt.tight_layout()
```

```python
# Turnover risk by tenure bucket
print('=== Attrition Rate by Tenure Bucket ===')
risk = df.groupby('TenureBucket')['is_terminated'].agg(['count', 'mean']).round(3)
risk.columns = ['Headcount', 'Attrition Rate']
print(risk)

fig, ax = plt.subplots(figsize=(8, 4))
risk['Attrition Rate'].plot(kind='bar', ax=ax, title='Turnover Risk by Tenure Bucket')
ax.set_ylabel('Attrition Rate')
plt.tight_layout()
```

```python
# Retirement risk (age 55+)
print('=== Retirement Risk (Age >= 55) ===')
near_retirement = df[df['Age'] >= 55]
print(f'Employees aged 55+: {len(near_retirement)} ({len(near_retirement)/len(df):.1%})')
print(f'\nBy department:')
print(near_retirement['DepartmentType'].value_counts())

fig, ax = plt.subplots(figsize=(8, 4))
near_retirement['DepartmentType'].value_counts().plot(kind='bar', ax=ax, title='Near-Retirement Employees by Department')
ax.set_ylabel('Count')
plt.tight_layout()
```
