# 04 - Performance Analytics
Analyze performance score distributions, correlations with tenure/age, and department-level trends.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_parquet('data/processed/workforce_clean_base.parquet')
print(f'Performance score distribution:')
print(df['Performance Score'].value_counts().sort_index())
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
df['Performance Score'].value_counts().sort_index().plot(kind='bar', ax=axes[0], title='Performance Score Distribution')
df.boxplot(column='Age', by='Performance Score', ax=axes[1])
axes[1].set_title('Age by Performance Score')
df.boxplot(column='TenureYears', by='Performance Score', ax=axes[2])
axes[2].set_title('Tenure by Performance Score')
plt.tight_layout()
```

```python
print('\n=== Performance by Department ===')
pd.crosstab(df['DepartmentType'], df['Performance Score'], normalize='index').round(3)
```

```python
print('=== Performance vs Attrition ===')
df['is_terminated'] = df['EmployeeStatus'].str.lower().isin(['terminated', 'voluntary termination', 'involuntary termination']).astype(int)
print(df.groupby('Performance Score')['is_terminated'].agg(['count', 'mean']).round(3))

fig, ax = plt.subplots(figsize=(8, 4))
df.groupby('Performance Score')['is_terminated'].mean().plot(kind='bar', ax=ax, title='Attrition Rate by Performance Score')
ax.set_ylabel('Attrition Rate')
plt.tight_layout()
```
