# 01 - Data Exploration
Initial exploration of the employee dataset: shape, dtypes, missing values, distributions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('data/raw/employee_data.csv')
print(f'Shape: {df.shape}')
print(f'\nColumns:\n{df.dtypes}\n')
print(f'\nMissing:\n{df.isnull().sum()}\n')
print(f'\nSummary:\n{df.describe(include="all")}')
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
df['EmployeeStatus'].value_counts().plot(kind='bar', ax=axes[0,0], title='Employee Status')
df['GenderCode'].value_counts().plot(kind='bar', ax=axes[0,1], title='Gender')
df['DepartmentType'].value_counts().plot(kind='bar', ax=axes[1,0], title='Department')
df['Performance Score'].value_counts().plot(kind='bar', ax=axes[1,1], title='Performance Score')
plt.tight_layout()
```
