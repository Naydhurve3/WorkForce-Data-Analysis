# 03 - Diversity & Inclusion
Analyze workforce demographics: gender balance, age diversity, generation mix, and representation by department.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_parquet('data/processed/workforce_clean_base.parquet')
print(f'Total employees: {len(df)}')
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
df['GenderCode'].value_counts().plot(kind='pie', autopct='%1.1f%%', ax=axes[0], title='Gender Distribution')
df['Generation'].value_counts().plot(kind='bar', ax=axes[1], title='Generation Distribution')
df['Age'].plot(kind='hist', bins=20, ax=axes[2], title='Age Distribution')
axes[2].set_xlabel('Age')
plt.tight_layout()
```

```python
print('\n=== Gender Representation by Department ===')
pd.crosstab(df['DepartmentType'], df['GenderCode'], normalize='index').round(3)
```

```python
print('=== Demographic Summary Statistics ===')
print(f'Mean age: {df["Age"].mean():.1f}')
print(f'Gender ratio (F/T): {df["GenderCode"].value_counts(normalize=True).get("Female", 0):.1%}')
if 'RaceDesc' in df.columns:
    print(f'\nRace/Ethnicity distribution:')
    print(df['RaceDesc'].value_counts(normalize=True).round(3))
```
