```python
import pandas as pd

# Load your dataset
df = pd.read_csv(r"../data/processed/cleaned2.csv")  # Replace with your actual file path if needed

# Display rows where DOB or Age is missing
missing_dob_age = df[df['DOB'].isna() | df['Age'].isna()]

# Show summary
print("🔍 Number of missing values:")
print("DOB missing:", df['DOB'].isna().sum())
print("Age missing:", df['Age'].isna().sum())

# Display the first few rows with missing DOB or Age
print("\n📋 Sample rows with missing DOB or Age:")
print(missing_dob_age[['EmpID', 'DOB', 'Age', 'SeniorityLevel', 'JobFamily']].head(10))
```

```python
import pandas as pd

# 🔹 Load the dataset
file_path = r"../data/processed/cleaned3.csv"
df = pd.read_csv(file_path)

# 🔍 Basic dataset structure
print("🔹 Dataset Shape (rows, columns):", df.shape)
print("\n🔹 Column Names:\n", df.columns.tolist())

# 🔍 Info about datatypes and non-null counts
print("\n🔹 Dataset Info:\n")
print(df.info())

# 🔍 Check for missing values
missing = df.isnull().sum()
missing_percent = (missing / len(df)) * 100
missing_data = pd.DataFrame({
    "Missing Values": missing,
    "Missing (%)": missing_percent
})
missing_data = missing_data[missing_data["Missing Values"] > 0]
print("\n🔍 Missing Values Summary:\n")
print(missing_data)

# 🔍 First 5 rows preview
print("\n🔹 First 5 Rows:\n")
print(df.head())

# 🔍 Basic stats for numerical columns
print("\n🔹 Summary Statistics for Numeric Columns:\n")
print(df.describe(include='number').T)

# 🔍 Basic stats for categorical columns
categorical_cols = df.select_dtypes(include='object').columns
print("\n🔹 Unique Values in Categorical Columns:\n")
for col in categorical_cols:
    print(f"{col} ({df[col].nunique()} unique): {df[col].unique()[:5]}")
```
