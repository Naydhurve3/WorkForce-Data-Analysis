```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
file_path = r"../../data/processed/cleaned3.csv"
df = pd.read_csv(file_path)

# Handle missing values in classification-related columns
df['EmployeeType'] = df['EmployeeType'].fillna('Unknown')
df['EmployeeClassificationType'] = df['EmployeeClassificationType'].fillna('Unknown')
df['SeniorityLevel'] = df['SeniorityLevel'].fillna('Unknown')
```

Employee Count by Employee Type

```python
type_counts = df['EmployeeType'].value_counts()

plt.figure(figsize=(8, 5))
sns.barplot(x=type_counts.index, y=type_counts.values, palette='Blues_r')
plt.title("Employee Count by Type")
plt.xlabel("Employee Type")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Employee Count by Classification Type

🟢 Purpose: Highlights structural employment categories (e.g., Permanent, Probationary, Consultant, etc.).

```python
class_counts = df['EmployeeClassificationType'].value_counts()

plt.figure(figsize=(8, 5))
sns.barplot(x=class_counts.index, y=class_counts.values, palette='Greens_d')
plt.title("Employee Count by Classification Type")
plt.xlabel("Employee Classification")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Seniority Level by Employee Type

🟢 Purpose: Understands how roles are distributed across types — are contractors mostly junior? Are full-timers spread across all levels?

```python
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='EmployeeType', hue='SeniorityLevel', palette='Set2')
plt.title("Seniority Level by Employee Type")
plt.xlabel("Employee Type")
plt.ylabel("Number of Employees")
plt.legend(title='SeniorityLevel', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Classification Type by Employee Type

🟢 Purpose: Explores how classification overlaps with employment type (e.g., Contract → Consultant or Probationary).

```python
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='EmployeeType', hue='EmployeeClassificationType', palette='Pastel1')
plt.title("Classification Type by Employee Type")
plt.xlabel("Employee Type")
plt.ylabel("Number of Employees")
plt.legend(title='Classification Type', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

📈 What We’re Analyzing and Why

| **Metric**                       | **Why It’s Used**                                                  |
| -------------------------------- | ------------------------------------------------------------------ |
| `EmployeeType`                   | To split workforce by employment type for planning and budgeting   |
| `EmployeeClassificationType`     | To identify contractual/legal status (e.g., Permanent, Consultant) |
| `SeniorityLevel`                 | To assess distribution of skill and leadership levels              |
| Visual overlays (bar/countplots) | To uncover role-type imbalances and workforce structure patterns   |
