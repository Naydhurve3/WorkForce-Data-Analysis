🔍 Objective:

To analyze recruitment trends over time using StartDate, and understand hiring patterns by JobFamily and SeniorityLevel. This helps with workforce planning, identifying gaps, and supporting future hiring strategies.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
file_path = r"../../data/processed/cleaned3.csv"
df = pd.read_csv(file_path)

# Convert StartDate to datetime
df['StartDate'] = pd.to_datetime(df['StartDate'], errors='coerce')
```

```python
df['JoinYear'] = df['StartDate'].dt.year
df['JoinMonth'] = df['StartDate'].dt.to_period('M')  # Useful for monthly trend
```

Plot Hiring Trends Over Years

```python
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='JoinYear', order=sorted(df['JoinYear'].dropna().unique()), palette='Blues')
plt.title("Recruitment Trend by Year")
plt.xlabel("Year of Joining")
plt.ylabel("Number of Employees Hired")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Monthly Hiring Trend (Optional Granularity)

```python
monthly_trend = df['JoinMonth'].value_counts().sort_index()

plt.figure(figsize=(14, 5))
monthly_trend.plot(kind='line', color='green')
plt.title("Monthly Hiring Trend")
plt.xlabel("Join Month")
plt.ylabel("Number of Hires")
plt.grid(True)
plt.tight_layout()
plt.show()
```

JobFamily-wise Hiring Trend

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='JobFamily', order=df['JobFamily'].value_counts().index, palette='Set2')
plt.title("Hiring Count by Job Family")
plt.xticks(rotation=45)
plt.ylabel("Number of Employees Hired")
plt.tight_layout()
plt.show()
```

Seniority Level Distribution Among Recruits

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='SeniorityLevel', palette='coolwarm', order=df['SeniorityLevel'].value_counts().index)
plt.title("Hiring Distribution by Seniority Level")
plt.xlabel("Seniority Level")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

✅ What We Analyzed and Why

| Step | What We Analyzed         | Why It’s Important                                                                        |
| ---- | ------------------------ | ----------------------------------------------------------------------------------------- |
| 1️⃣  | Join year/month          | Understand recruitment peaks and dips over time                                           |
| 2️⃣  | JobFamily                | Identify which roles are being hired most/least                                           |
| 3️⃣  | SeniorityLevel           | Check if hires are skewed toward entry, mid, or senior roles                              |
| 4️⃣  | Monthly trend (optional) | Helpful for **time-based hiring** analysis, such as seasonal or project-based recruitment |


📈 Why These Graphs Were Chosen

| Graph                                   | Why We Used It                                                                           |
| --------------------------------------- | ---------------------------------------------------------------------------------------- |
| `countplot` (Year)                      | Ideal for discrete yearly hiring trends                                                  |
| `line plot` (Month)                     | Shows temporal hiring pattern and volatility                                             |
| `countplot` (JobFamily, SeniorityLevel) | Best for categorical comparisons — which departments and levels are being recruited more |


🧠 What We Analyzed and Why

| Analysis Area                         | Description                                            | Why It's Important                                               |
| ------------------------------------- | ------------------------------------------------------ | ---------------------------------------------------------------- |
| **Yearly Hiring Trend**               | Number of employees hired each year (from `StartDate`) | To spot growth phases, hiring freezes, or expansion cycles       |
| **Monthly Hiring Pattern** (Optional) | Hiring activity at a monthly level                     | To identify seasonal or project-based hiring cycles              |
| **JobFamily Distribution**            | Number of hires per job family or department           | To evaluate which departments are scaling and which are stagnant |
| **Seniority Level of New Hires**      | Count of new hires across different seniority levels   | To ensure a balanced workforce pipeline (junior, mid, senior)    |


📈 Visualizations Used and Their Purpose

| Visualization                | Column(s) Used   | Purpose                                          | Why This Graph Was Chosen                                 |
| ---------------------------- | ---------------- | ------------------------------------------------ | --------------------------------------------------------- |
| `countplot` (Yearly Hires)   | `JoinYear`       | Show annual hiring trends clearly                | Bar charts are intuitive for year-wise comparisons        |

| `countplot` (JobFamily)      | `JobFamily`      | Compare hiring volumes across roles/departments  | Highlights which departments are growing or shrinking     |
| `countplot` (SeniorityLevel) | `SeniorityLevel` | Understand hiring focus (entry/mid/senior roles) | Great for workforce structure and skill pipeline analysis |


✅ Key Business Insights You Can Extract

- Hiring Trends: Are we hiring consistently? Were there spikes due to business growth, or dips due to hiring freezes or budget cuts?

- Departmental Growth: Are we over-hiring in certain job families (e.g., IT, Operations)? Are any critical departments under-resourced?

- Seniority Balance: Are we recruiting enough experienced professionals or relying too heavily on entry-level talent?

- Forecasting: Historical hiring trends help predict future recruitment needs and prepare HR budgets accordingly.
